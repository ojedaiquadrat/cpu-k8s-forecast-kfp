```python

from minio import Minio

# MinIO client
minio_client = Minio(
    "minio.minio.svc.cluster.local:9000",  # adjust to your namespace
    access_key="admin",
    secret_key="password",
    secure=False
)

# List all buckets
print("Buckets in MinIO:")
buckets = minio_client.list_buckets()
for bucket in buckets:
    print("-", bucket.name, "(created on:", bucket.creation_date, ")")
```
