## Simple code to deploy

```python
from kubernetes import client, config

config.load_incluster_config()

namespace = "lstm-iqu"
name = "nhits-model-vanilla"

body = {
    "apiVersion": "serving.kserve.io/v1beta1",
    "kind": "InferenceService",
    "metadata": {
        "name": name,
        "namespace": namespace,
    },
    "spec": {
        "predictor": {
            "minReplicas": 1,
            "maxReplicas": 1,
            "serviceAccountName": "sa-private-mlflow",
            "model": {
                "modelFormat": {
                    "name": "mlflow"
                },
                "runtime": "kserve-mlserver-nhits-vanilla",
                "storageUri": "s3://mlflow/24/62e3e98b98484c3eba55ea5a6c779645/artifacts/model_artifacts",
                "resources": {
                    "requests": {
                        "cpu": "500m",
                        "memory": "2Gi"
                    },
                    "limits": {
                        "cpu": "2",
                        "memory": "6Gi"
                    }
                }
            }
        }
    }
}

api = client.CustomObjectsApi()

group = "serving.kserve.io"
version = "v1beta1"
plural = "inferenceservices"

try:
    api.get_namespaced_custom_object(
        group=group,
        version=version,
        namespace=namespace,
        plural=plural,
        name=name,
    )
    api.patch_namespaced_custom_object(
        group=group,
        version=version,
        namespace=namespace,
        plural=plural,
        name=name,
        body=body,
    )
    print(f"Patched InferenceService: {name}")
except client.exceptions.ApiException as e:
    if e.status == 404:
        api.create_namespaced_custom_object(
            group=group,
            version=version,
            namespace=namespace,
            plural=plural,
            body=body,
        )
        print(f"Created InferenceService: {name}")
    else:
        raise
```

# IMPROVED Simple deploy

**Replace the InferenceService if it already exists, otherwise create it.
 This ensures the same service name is reused, and the runtime/storageUri
 are applied cleanly. We also add a restart annotation to force a new
 revision if the spec is unchanged.**

```python
# Replace the inferenceservice if ti already exists

from kubernetes import client, config

config.load_incluster_config()

namespace = "lstm-iqu"
name = "nhits-model-vanilla"

body = {
    "apiVersion": "serving.kserve.io/v1beta1",
    "kind": "InferenceService",
    "metadata": {
        "name": name,
        "namespace": namespace,
    },
    "spec": {
        "predictor": {
            "minReplicas": 1,
            "maxReplicas": 1,
            "serviceAccountName": "sa-private-mlflow",
            "model": {
                "modelFormat": {
                    "name": "mlflow"
                },
                "runtime": "kserve-mlserver-nhits-vanilla",
                "storageUri": "s3://mlflow/24/62e3e98b98484c3eba55ea5a6c779645/artifacts/model_artifacts",
                "resources": {
                    "requests": {
                        "cpu": "500m",
                        "memory": "2Gi"
                    },
                    "limits": {
                        "cpu": "2",
                        "memory": "6Gi"
                    }
                }
            }
        }
    }
}

api = client.CustomObjectsApi()

group = "serving.kserve.io"
version = "v1beta1"
plural = "inferenceservices"


try:
    existing = api.get_namespaced_custom_object(
        group=group,
        version=version,
        namespace=namespace,
        plural=plural,
        name=name,
    )
    body["metadata"]["resourceVersion"] = existing["metadata"]["resourceVersion"]
    body["metadata"]["annotations"] = {"restart-timestamp": "2026-04-23T19:20:00Z"}
    api.replace_namespaced_custom_object(
        group=group,
        version=version,
        namespace=namespace,
        plural=plural,
        name=name,
        body=body,
    )
    print(f"Replaced InferenceService: {name}")
except client.exceptions.ApiException as e:
    if e.status == 404:
        api.create_namespaced_custom_object(
            group=group,
            version=version,
            namespace=namespace,
            plural=plural,
            body=body,
        )
        print(f"Created InferenceService: {name}")
    else:
        raise
```

# Pipeline integration Code

storage_uri point the location of the model to be deployed. Up to you the strategy you wanna use to point it.

## use the ipykernel

```python
from typing import NamedTuple

from kfp import dsl


@dsl.component(
    base_image="python:3.11",
    packages_to_install=["kubernetes"],
)
def deploy_nhits_direct_from_storage_uri(
    storage_uri: str,
    service_name: str = "nhits-model-vanilla",
) -> NamedTuple(
    "Outputs",
    [
        ("service_name", str),
        ("storage_uri", str),
        ("status", str),
    ],
):
    import json
    import time

    from kubernetes import client
    from kubernetes import config

    config.load_incluster_config()

    namespace = "lstm-iqu"

    body = {
        "apiVersion": "serving.kserve.io/v1beta1",
        "kind": "InferenceService",
        "metadata": {
            "name": service_name,
            "namespace": namespace,
        },
        "spec": {
            "predictor": {
                "minReplicas": 1,
                "maxReplicas": 1,
                "serviceAccountName": "sa-private-mlflow",
                "model": {
                    "modelFormat": {
                        "name": "mlflow",
                    },
                    "runtime": "kserve-mlserver-nhits-vanilla",
                    "storageUri": storage_uri,
                    "resources": {
                        "requests": {
                            "cpu": "500m",
                            "memory": "2Gi",
                        },
                        "limits": {
                            "cpu": "2",
                            "memory": "6Gi",
                        },
                    },
                },
            },
        },
    }

    api = client.CustomObjectsApi()
    group = "serving.kserve.io"
    version = "v1beta1"
    plural = "inferenceservices"

    try:
        api.get_namespaced_custom_object(
            group=group,
            version=version,
            namespace=namespace,
            plural=plural,
            name=service_name,
        )
        api.patch_namespaced_custom_object(
            group=group,
            version=version,
            namespace=namespace,
            plural=plural,
            name=service_name,
            body=body,
        )
        action = "patched"
    except client.exceptions.ApiException as exc:
        if exc.status == 404:
            api.create_namespaced_custom_object(
                group=group,
                version=version,
                namespace=namespace,
                plural=plural,
                body=body,
            )
            action = "created"
        else:
            raise

    final_obj = None
    for _ in range(90):
        current = api.get_namespaced_custom_object(
            group=group,
            version=version,
            namespace=namespace,
            plural=plural,
            name=service_name,
        )
        conditions = current.get("status", {}).get("conditions", [])

        ready_condition = next(
            (cond for cond in conditions if cond.get("type") == "Ready"),
            None,
        )
        if ready_condition and ready_condition.get("status") == "True":
            final_obj = current
            break

        failed_condition = next(
            (
                cond
                for cond in conditions
                if cond.get("status") == "False"
                and cond.get("reason") == "RevisionFailed"
            ),
            None,
        )
        if failed_condition and failed_condition.get("message"):
            raise RuntimeError(
                f"InferenceService became unhealthy: {failed_condition.get('message')}"
            )

        time.sleep(10)

    if final_obj is None:
        final_obj = api.get_namespaced_custom_object(
            group=group,
            version=version,
            namespace=namespace,
            plural=plural,
            name=service_name,
        )
        raise RuntimeError(
            f"InferenceService '{service_name}' did not become Ready within the timeout. "
            f"Last status: {json.dumps(final_obj.get('status', {}), indent=2)}"
        )

    print(
        json.dumps(
            {
                "action": action,
                "service_name": service_name,
                "storage_uri": storage_uri,
                "status": "Ready",
            },
            indent=2,
        )
    )

    return (service_name, storage_uri, "Ready")


@dsl.pipeline(
    name="nhits-direct-deployment-storage-uri",
    description="Deploy an NHITS MLflow model to KServe using a direct storage URI.",
)
def deploy_nhits_direct_storage_uri_pipeline(
    storage_uri: str = "s3://mlflow/24/62e3e98b98484c3eba55ea5a6c779645/artifacts/model_artifacts",
    service_name: str = "nhits-model-vanilla",
):
    deploy_nhits_direct_from_storage_uri(
        storage_uri=storage_uri,
        service_name=service_name,
    ).set_caching_options(False)


if __name__ == "__main__":
    import kfp

    client = kfp.Client()
    client.create_run_from_pipeline_func(
        deploy_nhits_direct_storage_uri_pipeline,
        arguments={
            "storage_uri": "s3://mlflow/24/62e3e98b98484c3eba55ea5a6c779645/artifacts/model_artifacts",
            "service_name": "nhits-model-vanilla",
        },
        experiment_name="cpu-forecasting",
    )
```
