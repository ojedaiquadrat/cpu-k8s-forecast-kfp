```python
import json
import time
import random
from datetime import datetime
from kafka import KafkaProducer

producer = KafkaProducer(
    bootstrap_servers=["kafka.apache-kafka.svc.cluster.local:9092"],
    value_serializer=lambda v: json.dumps(v).encode("utf-8")
)

topic = "dummy-windowed"

print(f"Producing 5-value windows to topic: {topic}")

def generate_value():
    return round(random.uniform(0.1, 1.0), 6)

while True:
    # Create a window of 5 values with timestamps
    window = []
    for _ in range(5):
        window.append([
            {
                "timestamp": datetime.utcnow().isoformat(),
                "value": generate_value()
            }
        ])

    message = {"inputs": window}

    # Send to Kafka
    producer.send(topic, value=message)
    print("Sent:", json.dumps(message, indent=2))

    # Wait 5 seconds before next window
    time.sleep(5)
```
