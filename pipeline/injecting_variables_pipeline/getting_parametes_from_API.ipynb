```python
from kfp import compiler, dsl


@dsl.component(
    base_image="docker.io/jhofydu/pytorch-kfp:v1.0.0",
    packages_to_install=["minio==7.1.14"],
)
def get_data_batch(exp_id: str, status_url: str, data_url: str) -> None:
    print(f"exp_id={exp_id}")
    print(f"status_url={status_url}")
    print(f"data_url={data_url}")


@dsl.pipeline(
    name="sunrise-parameter-injection-demo",
    description="Demo pipeline that verifies runtime parameter injection in Kubeflow.",
)
def sunrise(
    exp_id: str = "",
    status_url: str = "",
    data_url: str = "",
) -> None:
    get_data_batch(
        exp_id=exp_id,
        status_url=status_url,
        data_url=data_url,
    ).set_caching_options(False)


compiler.Compiler().compile(
    pipeline_func=sunrise,
    package_path="injecting_parameter_test.yaml",
)
```
